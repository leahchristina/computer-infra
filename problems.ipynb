## Problem 1: Downloading Stock Market Data

In problem 1, my function must download stock data for 5 companies - NVIDIA, Broadcom, Arm, AMD and IMB, get 5 days of hourly data and save and download the data to a csv file which will also be written into a data folder in this respository, with a timestamped filename. 

Imports required for this function:

```python
# Downloading the data
import yfinance as yf

# Stores the data
import pandas as pd

# For folders
import os

# For timestamps
from datetime import datetime
```

The Download Function:

This creates a list of stock symbols, downloads 5 days of hourly data and creates a data folder if one does not already exist. It generates a timestamped filename and saves the downloaded data as a csv file.

```python
def get_data():
    
    # Stock symbols
    tickers = [
        "NVDA",
        "AVGO",
        "ARM",
        "AMD",
        "IBM"
    ]

    # Download data
    data = yf.download(
        tickers=tickers,
        period="5d",
        interval="1h",
        group_by="ticker"
    )

    # Create data folder if needed
    os.makedirs("data", exist_ok=True)

    # Create timestamp
    timestamp = datetime.now().strftime("%Y%m%d-%H%M%S")

    # Create filename
    filename = f"data/{timestamp}.csv"

    # Save data
    data.to_csv(filename)

    print(f"Data saved to: {filename}")

    return data
```

Running the Function

Downloading the market data and saving a copy.

```python
stock_data = get_data()

stock_data.head()
```

## Problem 2: Plotting the Data

As seen in Problem 1, the stock prices of different companies can vary signicantly. To make comparisons fairer on a plot, it is better to calculate the percentage change in each company's closing price relative to the first recorded value. This allows all companies to start at the same level and makes it easier to compare their relative performance over time rather than comparing the raw share prices. 

```python
# Required imports for this problem
import matplotlib.pyplot as plt
from pathlib import Path
```

Creating the Plot Function

The function below:
1. Finds the most recent csv file in the data folder.
2. Loads the stock market data.
3. Calculates the % change in the closing price for each company.
4. Creates a line chart & saves the chart into a plots folder.
5. Returns the percentage change data.

```python
def plot_data():

    # Find the most recent CSV file
    data_folder = Path("data")

    latest_file = max(
        data_folder.glob("*.csv"),
        key=lambda x: x.stat().st_mtime
    )

    print(f"Loading: {latest_file}")

    # Read the CSV file
    data = pd.read_csv(
        latest_file,
        header=[0, 1],
        index_col=0,
        parse_dates=True
    )

    # Stock symbols
    tickers = ["NVDA", "AVGO", "ARM", "AMD", "IBM"]

    # DataFrame to store percentage changes
    pct_change_df = pd.DataFrame(index=data.index)

    # Calculate percentage change
    for ticker in tickers:

        close_prices = data[(ticker, "Close")]

        first_price = close_prices.iloc[0]

        pct_change = (
            (close_prices - first_price)
            / first_price
        ) * 100

        pct_change_df[ticker] = pct_change

    # Create plots folder if needed
    os.makedirs("plots", exist_ok=True)

    # Create timestamped filename
    timestamp = datetime.now().strftime("%Y%m%d-%H%M%S")

    plot_file = f"plots/{timestamp}.png"

    # Create chart
    plt.figure(figsize=(12, 6))

    for ticker in tickers:
        plt.plot(
            pct_change_df.index,
            pct_change_df[ticker],
            label=ticker,
            linewidth=2
        )

    # Get date range from the data
    start_date = pct_change_df.index.min().strftime("%Y-%m-%d")
    end_date = pct_change_df.index.max().strftime("%Y-%m-%d")

    plt.title(
    f"Percentage Change in Share Price ({start_date} to {end_date})"
)

    plt.xlabel("Date and Time")
    plt.ylabel("Percentage Change (%)")

    plt.legend()

    plt.grid(True)

    plt.tight_layout()

    plt.savefig(plot_file)

    plt.show()

    print(f"Plot saved to: {plot_file}")

    return pct_change_df
```

```python
performance_data = plot_data()
```
